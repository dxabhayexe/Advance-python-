```python
##inheritance
#Multiple

class A:
    def showA(self):
        print("showA function")
class B(A):
    def showB(self):
        print("showB Function")
class C(B):
    def showC(self):
        print("showc Function")
obj=C()
obj.showA()
obj.showB()
obj.showC()
```

```python
# Hierarchical

class D:
    def showD(self):
        print("showD function")
class A(D):
    def showA(self):
        print("showA Function")
class B(D):
    def showB(self):
        print("showB Function")
class C(D):
    def showC(self):
        print("showC Function")
obj1=A()
obj1.showA()
obj1.showD()


obj2=B()
obj2.showB()
obj2.showD()
```

```python
#overriding

class parent:
   def bike(self):
    print("splender +")
class child(parent):
  def bike(self):
    print("HB")
obj=child()
obj.bike()
```
