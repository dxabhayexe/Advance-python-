```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
Data ={
    'Name':['A','B','C','D','E'],
    'Maths':[78, 67, 90, 56, 88],
    'Science':[85, 2,95 , 60, 84],
    'English':[88,70 , 92, 58, 86],
}
df=pd.DataFrame(Data)
print(df)
```

```python
marks = df[["Maths","Science","English"]].values
Average_marks = np.mean(marks, axis=1)
df["Average Marks"] = Average_marks
print(df)
```

```python
def grades(avg):
  if avg>=90:
    return "A"
  elif avg>=75:
    return "B"
  elif avg>=60:
    return "C"
  else:
    return 'D'
df['grades']=df["Average Marks"].apply(grades)
df
```

```python
plt.figure()
plt.bar(df['Name'],df['Average'])
plt.xlabel('students')
plt.ylabel("Average Marks")
plt.title("Average Marks of Students")
plt.show()
```

```python
#Average marks per subject

df[["Math","Science","English"]].mean().plot(kind="bar")
plt.title("Average Marks per Subject")
plt.ylabel(" Marks")
plt.show()
```
