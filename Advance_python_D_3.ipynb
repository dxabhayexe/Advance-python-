```python
#write data to.csv file

import csv

with open("data.csv", "w", newline="") as f:
    w=csv.writer(f)
    w.writerow(["ENO","ENAME","ESALARY","EADDR"])
    n=int(input("Enter no of records you want to enter:"))
    for i in range(n):
        eno=int(input("Enter employee number:"))
        ename=input("Enter employee name:")
        esalary=int(input("Enter employee salary:"))
        eaddr=input("Enter employee address:")
        w.writerow([eno,ename,esalary,eaddr])

    print("data written successfully to csv file")
```

```python
#Reading data from csv file

import csv
f=open("data.csv",'r')
r=csv.reader(f)
data=list(r)
print(data)
for line in data:
    for word in line:
        print(word,"\t",end=" ")
        print()
```

```python
import numpy as np

students=np.array([
    [85,90,78],
    [60,55,70],
    [95,92,88],
    [40,50,45],
    [72,68,75],

])
students
```

```python
avg_marks=np.mean(students,axis=1)
print("Average marks of each student:",avg_marks)
```

```python
avg_marks=np.mean(students,axis=0)
print("Average marks of each student:",avg_marks)
```

```python
topper = np.argmax(avg_marks)
print("Topper student :",topper)
```

```python
failed_students = np.where(avg_marks < 50)
print("Failed students:",failed_students[0]+1)
```

```python
for i,avg in enumerate(avg_marks):
    status = "Pass" if avg > 50 else "Fail"
    print(f"Student {i+1}: Average Marks = {avg:2f} |Status = {status}")
```
